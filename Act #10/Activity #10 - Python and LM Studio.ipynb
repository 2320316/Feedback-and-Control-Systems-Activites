```python
from openai import OpenAI

# Connect to LM Studio local server
client = OpenAI(
    base_url="http://localhost:1234/v1",
    api_key="lm-studio"  # any string works
)
```

```python
import keyboard
from langchain_community.chat_message_histories import ChatMessageHistory

# Initialize the history object to keep track of the discussion
history = ChatMessageHistory()
```

```python
MODEL_NAME = "mistralai/ministral-3-14b-reasoning"

while True:
    query = input("\nEnter your message (press Enter to continue, type 'exit' to quit): ").strip()

    # Exit condition
    if query.lower() == "exit":
        print("Thank you for using the AI assistant!")
        print("\nExiting...")
        break

    if not query:
        continue

    print(f"User: {query}")
    print("-+" * 50)

    # Save user message to history
    history.add_user_message(query)

    # Convert LangChain history to OpenAI format
    messages_with_history = []
    for msg in history.messages:
        if msg.type == "human":
            role = "user"
        elif msg.type == "ai":
            role = "assistant"
        else:
            role = "system"  # fallback
        messages_with_history.append({
            "role": role,
            "content": msg.content
        })

    # Stream AI response
    stream = client.chat.completions.create(
        model=MODEL_NAME,
        messages=messages_with_history,
        stream=True
    )

    full_ai_response = ""
    print("AI: ", end="", flush=True)

    for chunk in stream:
        delta = getattr(chunk.choices[0].delta, "content", None)
        if delta:
            print(delta, end="", flush=True)
            full_ai_response += delta

    print()

    # Save AI response to history
    history.add_ai_message(full_ai_response)

    print("*/" * 50)
```
